```python
# 문자열 판단하기
import re

re.match("Hello", "Hello, world!")
re.match("Python", "Hello, world!")
```

```python
# 문자열이 맨 앞에 오는지 맨 뒤에 오는지 판단하기 
# ^문자열 
# 문자열$
# re.search("패턴", "문자열")
re.search("^Hello", "Hello, world")
re.search("world!$", "Hello, world!")
```

```python
# 지정된 문자열이 하나라도 포함되는지 판단하기
re.match("hello|world", "hello")
```

```python
# 범위판단하기 
# [0-9]*
# [0-9]+

print(re.match("[0-9]*", "1234"))
print(re.match("[0-9]+", "1234"))
print(re.match("[0-9]+", "abcd"))
```

```python
print(re.match("a*b", "b"))
print(re.match("a+b", "b"))
print(re.match("a*b", "aab"))
print(re.match("a+b", "aab"))
```

```python
# 문자가 한 개만 있는지 판단하기 
# 문자?
# [0-9]?
print((re.match("abc?d", "abd")))
print((re.match("abc?d", "abcd")))
print((re.match("abc?d", "abcccd")))
print((re.match("ab[0-9]?c", "ab3c")))
print((re.match("ab.d", "abxd")))
```

```python
# 문자개수 판단하기 
# 문자{개수}
# (문자열){개수}

print(re.match("h{3}", "hhhello"))
print(re.match("h{3}", "hhhhello"))
print(re.match("hello{3}", "hellohellohelloworld"))
print(re.match("(hello){3}", "hellohellohelloworld"))
```

```python
print(re.match("[0-9]{3}-[0-9]{4}-[0-9]{4}", "010-1234-1234"))
print(re.match("[0-9]{3}-[0-9]{4}-[0-9]{4}", "010-1234-123"))
```

```python
# (문자){시작개수,끝개수}
# (문자열){시작개수,끝개수}
# [0-9]{시작개수,끝개수}
print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}", "02-123-1234"))
print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}", "010-1234-1234"))
print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}", "02-12-1234"))
```

```python
# 숫자와 영문 문자를 조합해서 판단하기 
# a-z
# A-Z
print(re.match("[a-zA-Z0-9]+", "eHllo1234"))
print(re.match("[A-Z0-9]+", "hello"))
```

```python
# 가-힣
re.match("[가-힣]+", "홍길동")
```

```python
# 특정 문자 범위에 포함되지 않는지 판단하기 
# [^범위]
# [^범위]+

print(re.match("[^A-Z]+", "Hello"))
print(re.match("[^A-Z]+", "hello"))
```

```python
# ^[범위]
# ^[범위]+
print(re.search("^[A-Z]+", "Hello"))
```
