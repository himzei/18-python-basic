```python
# 문자열 서식 지정자와 포매팅 사용하기 
```

```python
# 서식 지정자로 문자열 넣기 
# %s % "문자열"

"I am %s" % "james"
```

```python
name = "maria"
"I am %s" % name
```

```python
# 서식지정자로 숫자 넣기 
# "%d" % 숫자

"I am %d" % 20

b = 10

"I am %d" % b
```

```python
# 서식 지정자로 소수점 표현하기 
# "%f" % 숫자(실수)

"%f" % 2.3
```

```python
# "%.자리수f" % 숫자
print("%.2f" % 2.3)
print("%5.2f" % 2.3)
```

```python
# 서식 지정자로 문자열 정렬하기 
# %길이s

"%10s" % "python"
```

```python
# 왼쪽 정렬하기 
"%-10s" % "python"
```

```python
# 서식 지정자로 문자열 안에 값 여러개 넣기 
# "%d %s" % (숫자, 문자열)

print("Today is %d %s" % (3, "April"))
```

```python
# format 매서드 사용하기 
print("Hello, {0}".format("world"))
```

```python
# format 매서드로 값을 여러개 넣기 
print("Hello {0} {2} {1}".format("python", "Script", 3.6))
```

```python
# format 매서드로 같은 값을 여러개 넣기 
print("{0} {0} {1} {1}".format("Python", "Script"))
```

```python
# format 매서드에 인덱스 생략하기 
print("Hello {} {} {}".format("python", "Script", 3.6))
```

```python
# format매서드에서 인덱스 대신 이름 지정하기 
print("Hello {language} {version}".format(language="Python", version=3.6))
```

```python
# 문자열 포매팅에 변수를 그대로 사용하기 
language = "Python"
version = 3.6

print(f"Hello {language} {version}")
```

```python
# format 매서드로 문자열 정렬하기 
# "{인덱스:<10}".format("python")
print("{0:<10}".format("python"))
print("{0:>10}".format("python"))
print("{:>10}".format("python"))
```

```python
# 숫자 개수 맞추기
# "%0개수d" % 숫자
# "{인덱스:0개수d}".format(숫자)
print("%03d" % 1)
print("{:03d}".format(35))
```

```python
print("%08.2f" % 3.6)
print("{:08.2f}".format(150.37))
```

```python
# 채우기와 정렬을 조합해서 사용하기
print("{:0<10}".format(15))
print("{:0>10}".format(15))
```

```python
lis = map(int, input().split(";"))
for i in sorted(lis, reverse=True):
    print(f"{i:9,}")
```

```python
lis = list(map(int, input().split(";")))
lis.sort(reverse=True)
for i in lis:
    print(f"{i:9,}")
```
