```python
# 위치 인수를 사용하는 함수를 만들고 호출하기 
def print_numbers(a, b, c): 
    print(a)
    print(b)
    print(c)

print_numbers(10, 20, 30)
```

```python
# 언패킹 사용하기 
x = [10, 20, 30]
print_numbers(x) #=> ([10, 20, 30])
```

```python
x = [10, 20, 30]
print_numbers(*x) #=> ([10, 20, 30])
```

```python
# 가변인수 함수 만들기
# def 함수이름(*매개변수):
#     코드

def print_numbers(*args):
    for arg in args:
        print(arg)
```

```python
print_numbers(10)
print("=====")
print_numbers(10, 20, 30, 40)
```

```python
def print_numbers(*args):
    # print(args)
    # print(type(args))
    for arg in args:
        print(arg)

print_numbers(10, 20, 30)
```

```python
# 키워드 인수와 딕셔너리 언패킹 사용하기
# 함수(**딕셔너리)
def personal_info(name, age, address): 
    print("이름:", name)
    print("나이:", age)
    print("주소:", address)

x = {
    "name": "홍길동", 
    "age": 30, 
    "address": "서울시 용산구"
}

personal_info(**x)
# personal_info(name="홍길동", age=30, address="서울시 용산구")
```

```python
# 키워드 인수를 사용하는 가변 인수 함수 만들기
# def 함수이름(**매개변수):
#     코드

def personal_info(**kwargs):
    print(kwargs)
    print(type(kwargs))
    for kw, arg in kwargs.items():
        print(kw, ": ", arg, sep="")

personal_info(name="홍길동")
```

```python
personal_info(name="홍길동", age=30, address="서울시 용산구")
```

```python
def personal_info(**kwargs):
    for kw, arg in kwargs.items():
        print(kw, ": ", arg, sep="")

x = {
    "name": "홍길동", 
    "age": 30, 
    "address": "서울시 용산구",
    "hobby": ["축구", "농구", "게임"]
}

personal_info(**x)
```

```python
# 매개변수에 초깃값 지정하기
# def 함수이름(매개변수=값):
#     코드

def personal_info(name, age, address="비공개"): 
    print("이름:", name)
    print("나이:", age)
    print("주소:", address)


personal_info("홍길동", 30)
```

```python
personal_info("홍길동", 30, "서울시 이촌동")
```
