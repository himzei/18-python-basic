```python
# if 조건문의 동작 방식 알아보기 
if True:
    print("참")
else:
    print("거짓")

if False: 
    print("참")
else: 
    print("거짓")

if None:
    print("참")
else:
    print("거짓")
```

```python
# 조건문에 숫자 지정하기

if 0:
    print("참")
else: 
    print("거짓")

if 1: 
    print("참")
else: 
    print("거짓")

if 0x1f:
    print("참")
else: 
    print("거짓")

if 0b1000:
    print("참")
else:
    print("거짓")

if 13.5:
    print("참")
else: 
    print("거짓")
```

```python
# if조건문에 문자열 지정하기 
if "hello": 
    print("참")
else: 
    print("거짓")

if "":
    print("참")
else: 
    print("거짓")

if " ":
    print("참")
else: 
    print("거짓")
```

```python
# 조건식을 여러개 지정하기 
x = 0
y = 20 

if x == 10 & y:

    print("참")
else: 
    print("거짓")
```

```python
# 중첩 if조건문과 논리 연산자 

if x > 0:
    if x < 20:
        print("20보다 작은 양수입니다")
```

```python
if x > 0 and x < 20:
    print("20보다 작은 양수입니다")
```

```python
if 0 < x < 20: 
    print("20보다 작은 양수입니다")
```

```python
국어, 영어, 수학, 과학 = list(map(int, input().split()))

if 0 <= 국어 <= 100 and 0 <= 영어 <= 100 and 0 <= 수학 <= 100 and 0 <= 과학 <= 100:
    avg = (국어 + 영어 + 수학 + 과학) / 4
    if avg >= 80:
        print("합격")
    else:
        print("불합격")
else: 
    print("잘못된 점수")
```
