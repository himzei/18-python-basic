```python
# 재귀호출 사용하기 
def hello():
    print("Hello, world!")
    hello()

```

```python
hello()
```

```python
# 재귀호출에 종료 조건 만들기
def hello(count):
    if count == 0:
        return

    print("Hello world", count)

    count -= 1
    hello(count)
```

```python
hello(5)
```

```python
# 재귀호출로 팩토리얼 구하기

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)
```

```python
# 5 : 5 * factorial(4)
# 4 : 5 * 4 * factorial(3)
# 3 : 5 * 4 * 3 * factorial(2)
# 2 : 5 * 4 * 3 * 2 * factorial(1)
# 1 : 5 * 4 * 3 * 2 * 1
print(factorial(5))
```
