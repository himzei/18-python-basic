```python
# 덧셈함수 만들기
# def 함수이름(매개변수1, 매개변수2):
#     매개변수를 사용한 코드
def add(a, b):
    # a = 10, b = 20
    print(a + b)

result1 = add(10, 20)
print(result1)
```

```python
# 함수의 결과를 반환하기 
# def 함수이름(매개변수1, 매개변수2):
#     return 매개변수를 사용한 코드
def add(a, b):
    return a + b

result2 = add(10, 20)
print(result2)
```

```python
#1. 매개변수가 없고, 리턴되는 값도 없는 함수
#2. 매개변수가 없고, 리턴값은 있는 함수
#3. 매개변수가 있고, 리턴값은 없는 함수
#4. 매개변수가 있도, 리턴값도 있는 있는 함수
```

```python
#1. 매개변수 X, 리턴값 X
def say_hello():
    print("hello")

say_hello()
```

```python
#2. 매개변수 X, 리턴값은 O

def one():
    return 1

print(one())
```

```python
#3. 매개변수 O, 리턴값은 X
def add(a, b):
    # a = 10, b = 20
    print(a + b)
```

```python
#4. 매개변수 O, 리턴값은 O
def add(a, b):
    # a = 10, b = 20
    return a + b
```

```python
# return 으로 함수 중간에서 빠져나오기
def not_ten(a): 
    if a == 10: 
        return
    print(a, "입니다", sep="")
```

```python
not_ten(5)
```

```python
not_ten(10)
```

```python
# 함수에서 값을 여러개 반환하기 
def add_sub(a, b):
    return a + b
    return a - b

add_sub(10, 20)
```

```python
def add_sub(a, b):
    return a + b, a - b


result = add_sub(10, 20)
result
```

```python
x, y = add_sub(10, 20)
print(x)
print(y)
```

```python
# 함수의 호출 과정 알아보기
def mul(a, b):
    c = a * b 
    return c 

def add(a, b):
    c = a + b
    print(c)
    d = mul(a, b)
    print(d)


x = 10
y = 20
add(x, y)
```

```python
x, y = map(int, input().split())

def calc(num_a, num_b):
    add = num_a + num_b 
    sub = num_a - num_b 
    mul = num_a * num_b 
    div = num_a / num_b 
    return add, sub, mul, div

a, s, m, d = calc(x, y)
print("덧셈: {0}, 뺄셈: {1}, 곱셈: {2}, 나눗셈: {3}".format(a, s, m, d))
```
