```python
# 딕셔너리 표현식에 if조건문 사용하기 
x = {
    "a": 10, 
    "b": 20, 
    "c": 30, 
    "d": 40
}

for key, value in x.items():
    if value == 20:
        del x[key]

print(x)
```

```python
col, row = map(int, input().split())

matrix = []
for i in range(row): 
    matrix.append(list(input()))
```

```python
matrix
```

```python
print(*matrix, sep="\n")
```

```python
for i in range(row):
    for j in range(col):
        if matrix[i][j] == ".":
            count = 0
            for r in range(i-1, i+1+1):
                for c in range(j-1, j+1+1):
                    if 0 <= r < row and 0 <= c < col: 
                        if matrix[r][c] == "*": 
                            count += 1
            matrix[i][j] = count

print(*matrix, sep="\n")
```

```python
for i in matrix:
    for j in i:
        print(j, end="")
    print()
```
