```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기 
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

for x, y in a:
    print(x, y, end=" ")
```

```python
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]
for i in a: 
    for j in i:
        print(j, end=" ")
    print()
```

```python
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=" ")
    print()
```

```python
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

i = 0
while i < len(a): 
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
# while 반복문을 두 번 사용하기 
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

i = 0
while i < len(a):
    j = 0
    while j < len(a[i]):
        print(a[i][j], end=" ")
        j += 1
    print()
    i += 1

```

```python
col, row = map(int, input().split())
matrix = []
for i in range(row):
    matrix.append(list(input()))
```

```python
list(".**")
```

```python
matrix

```

```python
print(*matrix, sep="\n")
```

```python
for i in range(row): 
    for j in range(col): 
        if matrix[i][j] == ".":

            count = 0
            # count 개수를 추가를 해야함 
            for r in range(i-1, i+2):
                for c in range(j-1, j+2): 

                    if 0 <= r < row and 0 <= c < col:
                        if matrix[r][c] == "*":
                            count += 1

            matrix[i][j] = count
print(*matrix, sep="\n")
```
