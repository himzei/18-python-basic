```python
# 람다표현식으로 함수 만들기
def plus_ten(x): 
    return x + 10

plus_ten(1)
```

```python
# lambda 매개변수들: 식
lambda x: x + 10
```

```python
plus_ten2 = lambda x: x + 10
plus_ten2(2)
```

```python
# 람다표현식 자체를 호출하기 
# (lambda 매개변수 : 식)(인수들)
(lambda x: x+10)(2)
```

```python
# 람다표현식 안에서는 변수를 만들수 없다. 
(lambda x: y = 10; x + y)(1)
```

```python
y = 10
(lambda x: x + y)(3)
```
